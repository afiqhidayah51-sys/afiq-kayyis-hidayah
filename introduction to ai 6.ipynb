# Klasifikasi Minat Membeli Produk (Naive Bayes)

Tugas: Mengklasifikasikan apakah pelanggan Tertarik atau Tidak Tertarik menggunakan metode Naive Bayes.

```python
# Data training
data = [
    ["Murah", "Ada", "Baik", "Tertarik"],
    ["Murah", "Ada", "Baik", "Tertarik"],
    ["Sedang", "Ada", "Baik", "Tertarik"],
    ["Mahal", "Tidak Ada", "Buruk", "Tidak Tertarik"],
    ["Mahal", "Tidak Ada", "Buruk", "Tidak Tertarik"],
    ["Sedang", "Ada", "Baik", "Tertarik"],
    ["Sedang", "Tidak Ada", "Baik", "Tertarik"],
    ["Mahal", "Ada", "Buruk", "Tidak Tertarik"]
]

# Data baru
test = ["Sedang", "Ada", "Baik"]
```

```python
# Hitung prior
total = len(data)
tert_count = sum(1 for d in data if d[3] == "Tertarik")
tdk_count = sum(1 for d in data if d[3] == "Tidak Tertarik")

p_tert = tert_count / total
p_tdk = tdk_count / total

print("Prior Tertarik:", p_tert)
print("Prior Tidak Tertarik:", p_tdk)
```

```python
# Fungsi likelihood
def likelihood(feature_index, value, label):
    subset = [d for d in data if d[3] == label]
    count = sum(1 for d in subset if d[feature_index] == value)
    return count / len(subset)

# Hitung posterior
p_x_tert = p_tert
p_x_tdk = p_tdk

for i in range(3):
    p_x_tert *= likelihood(i, test[i], "Tertarik")
    p_x_tdk *= likelihood(i, test[i], "Tidak Tertarik")

print("\nPosterior Tertarik:", p_x_tert)
print("Posterior Tidak Tertarik:", p_x_tdk)
```

```python
# Hasil klasifikasi
if p_x_tert > p_x_tdk:
    print("\nPrediksi: TERTARIK")
else:
    print("\nPrediksi: TIDAK TERTARIK")
```
